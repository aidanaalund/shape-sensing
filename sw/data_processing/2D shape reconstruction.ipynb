```python
# generate synthetic data for testing of a cantilevered beam image that is segmented into the beam of equal width

import pandas as pd
from pathlib import Path
import numpy as np

path = Path("sw/sample_data/centerlines/pof_00000.csv")
df = pd.read_csv(path, header=None, names=["s_norm", "x_px", "y_px", "theta_rad", "curvature_per_m"])
# columns: s_norm, x_px, y_px, theta_rad, curvature_per_m
# 600 rows, clamp (s_norm=0) to tip (s_norm=1)

df.head()
```

```python
# 2D shape reconstruction plot
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 2))
plt.plot(df["x_px"], df["y_px"], label="Centerline", color="blue")
plt.title("Synthetic Cantilevered Beam Centerline")
plt.xlabel("X Position (pixels)")
plt.ylabel("Y Position (pixels)")
plt.xlim(0, 100)
plt.ylim(-10, 10)
plt.axhline(0, color='gray', linestyle='--', linewidth=0.5)
plt.legend()
plt.grid()
```
